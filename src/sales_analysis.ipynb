```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum as sum_amount

spark = SparkSession.builder.appName(
    "SalesAnalysis"
).getOrCreate()

df = spark.read.option("header", True).option(
    "inferSchema", True
).csv("data/sales.csv")

df.show()

sales_by_city = df.groupBy("city").agg(
    sum_amount("amount").alias("total_sales")
)

sales_by_city.show()
```
